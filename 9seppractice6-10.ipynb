```python
# function for to check input is palindrome or not:

def pali_drome(objects:str)-> bool:
    """This tells you a string palindrome or not
    args:
       objects: it must be string
    return:
        bool value
    """
    palindrome = ""
    for i in objects:
        palindrome = i + palindrome
    if palindrome == objects:
        return True
    return False
try:
    user_input = input("enter your string").lower()
    if pali_drome(user_input):
        print(f"{user_input} is a palindrome")
    else:
        print(f"{user_input} not a palindrome")
    
except Exception as e:
    print(e)
```

```python
# function for count_vowels in input:

def count_vowels(objects:str)->int:
    """This function helps to find vowels in string
    args:
       objects: it must be string
    returns:
        it gives you a integer value
    """
    vowels = ["a","e","i","o","u"]
    count = 0  
    for i in objects:
        if i in vowels:
            count+=1
    return count
if __name__ =="__main__":
    try:
        user_input = input("enter you input")
        print("number of vowels in this input:",count_vowels(user_input))
    except Exception as e:
    print(e)
```

```python
# function for find second_largest number in input:

def second_largest(objects: list)-> int:
    """ This function helps to find the second and first largest number in list
    args:
       objects: it should be list
    returns:
        it gives you a integer value
    """
    # lst = [1,22,33,44,221,2,3,4,5,50]
    if objects[0]>objects[1]:
        largest_number = objects[0]
        second_largest_number = objects[1]
    else:
        largest_number = objects[1]
        second_largest_number = objects[0]
    for i in objects[2::]:
        if i>largest_number:
            second_largest_number = largest_number
            largest_number = i
        elif i != largest_number and i>second_largest_number:
            second_largest_number = i
    return second_largest_number , largest_number
if __name__ == "__main__":
    try:
        user_input = input("enter you number: ")
        my_list = [int(x) for x in user_input.split(" ")]
        print("second_largest , largest_number is :",second_largest(my_list))
    except Exception as e:
        print("input must be number (give a number using spacebar)")
```

```python
# function for sum of all the digit of input: 

def addition_digit(num:int)->int:
    """This funciton calculates the sum of all digit in a given number
    args:
      num: it must be integer
    returns:
      it gives you a integer value representing the sum of the digit
    """
    if not isinstance(num,int):
        print("input must be number (decimal value also not consider)")
    sum_digit = 0
    while num>0:
        result = num%10
        sum_digit+=result
        num = num//10
    return sum_digit
if __name__ == "__main__":
    try:
        number = 455
        print(addition_digit(number))
    except TypeError as te:
        print(te)
    except Exception as e:
        print(e)
```

```python
# function for remove duplicate element:

def remove(objects:list)->list:
    """ This function helps to remove duplicate element from the list
    args:
        objects: it must be list
    returns:
        it gives you a list of unique number
    """
    if not isinstance(objects,list):
        raise TypeError("input must be list")
    my_list = []
    for i in objects:
        num = int(i)
        if num not in my_list:
            my_list.append(num)
    return my_list
try:
    user_input = input("enter your list")
    my_list = [int(x) for x in user_input.split(" ")]
    print(remove(my_list))
except Exception as e:
    print(e)
```
