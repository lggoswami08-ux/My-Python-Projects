```python
# 21. GCD using Recursion:
def GCD(num1:int,num2:int):
    """
    This function helps to find GCD of two numbers
    args: 
        num1: it must be integer.
        num2: it's also must be integer.
    return:
        it's give you a Greatest Common Divisor of two number.
    """
    if not isinstance(num1,int) or not isinstance(num2,int):
        raise TypeError("input must be numeric value")
    num1 = abs(num1)
    num2 = abs(num2)
    if num2==0: 
        return num1
    return GCD(num2,num1%num2)
if __name__ == "__main__":
    try: 
        user_input_num1 = 45
        user_input_num2 = 12
        print(GCD(user_input_num1,user_input_num2))
    except TypeError as TE:
        print(TE)
```

```python
# 22. Program to Convert Decimal to Binary Without Using bin().
def binary(objects:int):
    """ 
    This function gives you a binary form of a number.
    args:
        objects: it must be a integer
    """
    if not isinstance(objects,int):
        raise TypeError("input must be numeric value")
    decimal = 5
    binary = ""
    while decimal>0:
        reminder = decimal%2
        decimal = decimal//2
        binary = str(reminder)+binary
    return binary
if __name__=="__main__":
    try:
        user_input = 5
        print(f"binary form of {user_input}:",binary(user_input))
    except Exception as e:
        print(e)
    
```

```python
# 23. find intersection of two list.
def intersection(object1:list,object2:list)->list:
    """
    This function gives you a common element btw two list.
    args:
        object1: it must be list
        object2: it also must be a list
    returns:
        list of common element
    """
    if not isinstance(object1,list) or not isinstance(object2,list):
        raise TypeError("input must in the form of list")
    list3 = []
    for i in object1:
        if i in object2 and i not in list3:
            list3.append(i)
    return list3
if __name__=="__main__":
    try:
        user_list1 = [int(x) for x in input("enter your number using spacebar:").split(" ")]
        user_list2 = [int(x) for x in input("enter your number using spacebar:").split(" ")]
        print("common element between two list:",intersection(user_list1,user_list2))
    except TypeError as TE:
        print(TE)
    except ValueError as ve:
        print("input must be number")
    except Exception as e:
        print(e)
```

```python
# 24. Remove all whitespace from A String:
def Remove(text:str):
    """
    This function helps to remove all whitespace from the string
    args:
        text: it must be string
    """
    if not isinstance(text,str):
        raise TypeError("input must in form of string")
    result = ""
    prev_space = False 
    for i in text.strip():
        if not i.isspace():
            result = result+i
            prev_space = False
        else:
            if prev_space == False:
                result = result + " "
                prev_space = True
    return result
if __name__ == "__main__":
    try: 
        user_input = "       \n     this     \r     is     python"
        print(Remove(user_input))
    except TypeError as te:
        print(te)
```

```python
# 25. Find Sum of List Using Recursion:
def list_sum(my_list:list):
    """
    This function helps to sum all element of the list
    args:
        My_list: it must be list.
    retrun:
        sum of all element of the list 
    """
    if not isinstance(my_list,list):
        raise TypeError("input must be list")
    if len(my_list)==0:
        return 0
    else:
        return my_list[0]+list_sum(my_list[1::])
if __name__ =="__main__":
    try:
        user_input = [10,20,30,40]
        print(list_sum(user_input))
    except TypeError as te:
        print(te)
    except ValueError as ve:
        print(ve)
```
