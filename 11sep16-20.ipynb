```python
# 16. Calculate the power of number without using **Operator:
def power_of_number(base_value:int,power_value:int)->int:
    """Calculate power of a number without using**operator.
    args:
        base_value: input must be integer
        power_value: input must be non negative number
    return:
        int: base_value raise to the power power_value
    """
    if not isinstance(base_value,int):
        raise TypeError("input must be number")
    if power_value<0:
        raise ValueError("negative value can't consider as power")
    common_number = 1
    for i in range(power_value):
        common_number = base_value*common_number
    return common_number
if __name__=="__main__":
    try:
        base_value = 5
        power_value = 2
        print(power_of_number(base_value,power_value))
    except (TypeError,ValueError) as TV:
        print(TV)
    except Exception as e:
        print(e)
```

```python
# 17. Program to Generate List of Even Numbers:
def even_num(objects:int)->list:
    """
    Function to generate a list of even numbers
    args:
         objects : it must be integers
    returns:
        list: list of even number form 1 to objects
    """
    if not isinstance(objects,int):
        raise TypeError("input must be numeric value")
    if objects<2:
        raise ValueError("input must be greater than 2 or equal to 2")
    empty_list=[i for i in range(2,objects+1,2)]
    return empty_list
if __name__ == "__main__":
    try:
        user_input = 5
        print(even_num(user_input))
    except (TypeError,ValueError) as TV:
        print(TV)
    except Exception as e:
        print(e)
    
```

```python
# 18. Binary search implimentation in python.
def binary_search(sorted_list:list,target:int)->int:
    """
    Function to perform  binary search on a sorted list.
    args: 
        sorted_list : it must be list
        target: it must be integer value
    returns:
        int: Index of the target element if found, otherwise -1.
    """
    if not isinstance(sorted_list,list):
        raise TypeError("input must in the form of list")
    if not isinstance(target,int):
        raise TypeError("target must be a number/integer")
    low = 0
    high = len(sorted_list)-1
    while low<=high:
        mid = (low+high)//2
        if sorted_list[mid]==target:
            return mid
            break
        elif target<sorted_list[mid]:
            high = mid-1
        else:
            low = mid+1
    else:
        return -1
if __name__=="__main__":
    try:
        user_input = [2,8,45,12,16,23,38,67,56,78,90,89,100,95]
        user_input.sort()
        target = 90
        print(binary_search(user_input,target))
    except (TypeError,ValueError) as TV:
        print(TV)
        
```

```python
# 19. Reverse words in a sentence.
def reverse_word_function(sentence:str)->str:
    """ 
    Reverse the order of words in a sentence.
    args:
        Sentence: it must be string
    returns:
        str: sentence with words reverse.
    """
    if not isinstance(sentence,str):
        raise TypeError("input must be string")
    word = ""
    list1 = []
    for i in sentence:
        if i != " ":
            word = word+i
        else:
            list1.append(word)
            word = ""
    list1.append(word)
    reverse_sentence = ""
    for i in range(len(list1)-1,-1,-1):
        reverse_sentence = reverse_sentence + list1[i]
        if i != 0:
            reverse_sentence = reverse_sentence + " "
    print(reverse_sentence)
if __name__ == "__main__":
    try:
        user_input = "python is fun"
        reverse_word_function( user_input)
    except TypeError as e:
        print(e)
        
```

```python
# 20. Program to find GCD(Greatest Common Divisor) Using Euclidean Algorithm.
def GCD(num1:int,num2:int):
    """
    This function used to find GCD of two numbers
    args:
        num1: it must be integers
        num2: it must be integers
    returns:
        it's gives you a Greater Common Divisor of two numbers
    """
    if not isinstance(num1,int) or not isinstance(num2,int):
        raise TypeError("number must integer")
    num1 = abs(num1)
    num2 = abs(num2)
    while num2!= 0:
        num1,num2 = num2,num1%num2
    return num1
if __name__=="__main__":
    try:
        num1_input = 20
        num2_input = 30
        print(f"Greater Common Divisor of {num1_input} and {num2_input}:",GCD(num1_input,num2_input))
    except TypeError as t:
        print(t)
```
