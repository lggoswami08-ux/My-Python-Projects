```python
# 26. Bubble sort in python:

def sorted(objects:list)->list:
    """
    This function helps you to sort a list
    objects: 
        it must be a list
    returns:
        a sorted list
    """
    if not isinstance(objects,list):
        raise TypeError("input must be a list")
    length = len(objects)
    for i in range(length):
        for j in range(0,length-i-1):
            if objects[j]>objects[j+1]:
                objects[j],objects[j+1]=objects[j+1],objects[j]
    return objects
if __name__=="__main__":
    try:
        user_input = input("Enter your number(using spacebar)")
        empty_list = [int(i) for i in user_input.split(" ")]
        print(sorted(empty_list))
    except TypeError as te:
        print(te)
```

```python
# 27. merge two sorted list:

def merge_sort(user_input1:list,user_input2:list)->list:
    """
    This function gives you a merge of sorted list
    args:
        user_input1 = it must be in a form of list
        user_input2 = also in a form of a list
    returns:
        a sorted list
    """
    if not isinstance(user_input1,list) or not isinstance(user_input2,list):
        raise TypeError("input must be in a form of list")
    merge_list = []
    i=j=0
    while i<len(user_input1) and j<len(user_input2):
        if user_input1[i]<=user_input2[j]:
            merge_list.append(user_input1[i])
            i+=1
        else:
            merge_list.append(user_input2[j])
            j+=1
    while i<len(user_input1):
        merge_list.append(user_input1[i])
        i+=1
    while j<len(user_input2):
        merge_list.append(user_input2[j])
        j+=1
    return merge_list
if __name__=="__main__":
    try:
        user_list = input("enter your number (using spacebar)")
        a = [int(i) for i in user_list.split(" ")]
        a.sort()
        user_list2 = input("enter you sec numbers(using spacebar")
        b = [int(i) for i in user_list2.split(" ")]
        b.sort()
        print(merge_sort(a,b))
    except TypeError as tv:
        print(tv)
```

```python
# 28. check the list is sorted or not(True/False):

def check(objects:list)->bool:
    """
    This function gives you a bool value if list is sorted then its gives you a True otherwise False
    args:
        objects: it must be a list
    returns:
        a bool value
    """
    if not isinstance(objects,list):
        raise TypeError("input must be in a form of list")
    for i in range(len(objects)-1):
        if objects[i]>objects[i+1]:
            return False
    else:
        return True
if __name__ == "__main__":
    try: 
        user_input = input("enter your number (using spacebar)")
        a = [int(i) for i in user_input.split(" ")]
        if check(a):
            print(f"{a} it's  a sorted list")
        else:
            print(f"{a} it's not a sorted list")
    except TypeError as te:
        print(te)
```

```python
# 29. Count frequency of each word in a text file.

import string
def frequency()->dict:
    """
    This function helps to count a frequency of each word in a text file
    returns
        it gives you a dictionary
    """
    with open(r"C:\Users\lggos\OneDrive\Desktop\New Text Document.txt",mode = 'r') as file:
        text = file.read()
    pattern = str.maketrans("","",string.punctuation)
    clear_text = text.translate(pattern).lower().split()
    empty_dict = {}
    for i in clear_text:
        if i not in empty_dict:
            empty_dict[i]=1
        else:
            empty_dict[i]+=1
    return empty_dict
if __name__ == "__main__":
    try:
        print(frequency())
    except Exception as e:
        print(e)
```

```python
# 30. Find missing number in a list:
def missing(objects:list)->int:
    """
    This function gives you a missing number
    args:
        objects: it must be list
    returns:
        it gives you a missing number
    """
    if not isinstance(objects,list):
        raise TypeError("input must be list")
    for i in range(len(objects)-1):
        if objects[i+1]-objects[i]!=1:
            print("missing number is:",objects[i]+1)
    else:
         return None
if __name__=="__main__":
    try:
        user_input = [1,2,3,4,7]
        missing(user_input)
    except TypeError as te:
        print(te)
```
