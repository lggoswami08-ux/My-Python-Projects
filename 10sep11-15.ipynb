```python
# 11. factorial using recursion:
def factorial(objects:int)->int:
    """This function help to find the factorial of the number
    args:
       objects: it must be int
    returns:
        it gives you a factorial of number
    """
    if not isinstance(objects,int):
        raise TypeError("input must be number")
    if objects==0 or objects==1:
        return 1
    result = objects*factorial(objects-1)
    return result
if __name__=="__main__":
    try:
        user_input = 5
        print(f"factorial of {user_input}:",factorial(user_input))
    except TypeError as te:
        print(te)
    except Exception as e:
        print("non integer value not allowed")
```

```python
# 12. count occurrence of each characters:
def count(objects:str)->dict:
    """This funciton help to count the occurrence of the character
    args:
       objects: it must be string
    returns:
        it gives you a dictionary value
    """
    if not isinstance(objects,str):
        raise TypeError("input must be string")
    empty_dict = {}
    for i in objects.lower().replace(" ",""):
        if i not in empty_dict:
            empty_dict[i]=1
        else:
            empty_dict[i]+=1
    for char,count in empty_dict.items():
        print(char,count)
if __name__ == "__main__":
    try: 
        user_input = "this is"
        count(user_input)
    except TypeError as te:
        print(te)
    except Exception as e:
        print(e)
```
