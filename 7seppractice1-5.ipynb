```python
# reverse the string using for 
def reverse_string(objects:str)->string:
    """This function help to find reverse string
    args:
       objects: it must be string
    returns: 
        reverse string
    """
    if not isinstance(objects,str):
        raise TypeError("input must be string")
    reverse_string1 = ""
    for i in objects:
        reverse_string1 = i + reverse_string1
    return reverse_string1
if __name__=="__main__":
    try:
        print(reverse_string(hhhhh))
    except TypeError as Te:
        print(Te)
    except Exception as e:
        print(e)

    
```

```python
def largest_element(objects: list)->int:
    """This function help to find the largest number in the list
    args:
       objects: it must be list
    returns:
        largest element of the list
    """
    if not isinstance(objects,list):
        raise TypeError("input must be list")
    largest_element = objects[0]
    for i in objects:
        if i > largest_element:
            largest_element = i
    return largest_element
if __name__=="__main__":
    try:
        print(largest_element([111,222222,3333,44444]))
    except TypeError as Te:
        print(Te)
    except Exception as e:
        print(e)
```

```python
# function for checking number is prime or not 
def prime_or_not(num:int)->bool:
    """This function check the number is prime or not
    args:
       num: it must be interger
    retrun:
        bool value
    """
    if not isinstance(num,int):
        raise TypeError("input should be number")
    if num<2:
         raise ValueError(" less than 2 can't consider")
    for i in range(2,num):
        if num%i==0:
            return True
    return False
try:
    user_input = 12
    if prime_or_not((user_input)):
        print(f"{user_input} is not a prime number")
    else:
        print(f"{user_input} it is a prime number")
except (TypeError,ValueError) as TV:
    print(TV)
except Exception as e:
    print("input must be number")
```

```python
# find all prime numbers between 100 to 200
def prime_100_200(num1: int , num2: int):
    """this function tells you prime number between num1 and num2
    args:
       num1: it must be integer
       num2: also a integer
    return:
       it gives you a prime numbers
    """
    if not isinstance(num1,int):
        raise TypeError("input must be integers")
    if not isinstance(num2,int):
        raise TypeError("input must be integers")
    for num in range(num1,num2):
        for i in range(2,num):
            if num%i==0:
                break
        else:
            print(num,end=" ")
try:
    prime_100_200(400,500)
except TypeError as te:
    print(te)
except Exception as e:
    print(e)
        
```

```python
def factorial(objects:int):
    """This function help to find the factorial of number
    args:
       objects: it must be integer
    return:
       it's gives you a factorial of number
    """
    # user_input = input("enter you number")
    if not isinstance(objects,int):
        raise TypeError("Input must be integer")
    if objects<1 or objects == 0:
        raise ValueError("negative or 0 value not consider")
    factorial = 1
    for i in range(1,objects+1):
        factorial = factorial*i
    return factorial
try:
    user_input = input("input should be integer")
    num = int(user_input)
    print(factorial(num))
except TypeError as te:
    print("Input must be integer")
except ValueError as ve:
    print(ve)
except Exception as e:
    print("input must be integer")
```
